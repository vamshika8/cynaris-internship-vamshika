```python
import pandas as pd

# Create the first DataFrame
employees = pd.DataFrame({
    "employee_id": [1, 2, 3, 4],
    "name": ["Asha", "Rahul", "Priya", "Kiran"],
    "department_id": [101, 102, 101, 103]
})

# Create the second DataFrame
departments = pd.DataFrame({
    "department_id": [101, 102, 104],
    "department": ["Data Science", "HR", "Finance"]
})

# Inner join
inner_join = pd.merge(employees, departments, on="department_id", how="inner")
display(inner_join)
```

```python
# Left join
left_join = pd.merge(
    employees, departments,
    on="department_id", how="left"
)
print("LEFT JOIN")
display(left_join)

# Right join
right_join = pd.merge(
    employees, departments,
    on="department_id", how="right"
)
print("RIGHT JOIN")
display(right_join)

# Outer join
outer_join = pd.merge(
    employees, departments,
    on="department_id", how="outer"
)
print("OUTER JOIN")
display(outer_join)
```

```python
# Create a sales DataFrame
sales = pd.DataFrame({
    "region": ["North", "North", "South", "South",
               "East", "East", "North", "South"],
    "product": ["Laptop", "Phone", "Laptop", "Phone",
                "Laptop", "Phone", "Phone", "Laptop"],
    "sales": [50000, 20000, 45000, 25000,
              60000, 30000, 15000, 35000]
})

# Create a pivot table
pivot_table = pd.pivot_table(
    sales,
    values="sales",
    index="region",
    columns="product",
    aggfunc="sum",
    fill_value=0
)

display(pivot_table)
```

```python
# Create a wide-format DataFrame
wide_data = pd.DataFrame({
    "region": ["North", "South", "East"],
    "Laptop": [50000, 45000, 60000],
    "Phone": [35000, 25000, 30000]
})

print("Wide Format:")
display(wide_data)

# Convert wide data to long format
long_data = pd.melt(
    wide_data,
    id_vars=["region"],
    var_name="product",
    value_name="sales"
)

print("Long Format:")
display(long_data)
```

```python
# Chain multiple Pandas operations
result = (
    sales
    .query("sales >= 30000")
    .groupby("region", as_index=False)["sales"]
    .sum()
    .sort_values("sales", ascending=False)
)

display(result)
```
