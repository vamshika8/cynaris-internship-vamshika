```python
import pandas as pd
import numpy as np
from pathlib import Path

# Load the sales dataset
DATA_PATH = Path("../../north_sales.csv")
df = pd.read_csv(DATA_PATH)

# Display the original dataset
print("Original Dataset:")
display(df.head())

print("\nDataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
# Check missing values
print("Missing Values Before Cleaning:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate Rows Before Cleaning:")
print(df.duplicated().sum())

# Check data types
print("\nData Types:")
print(df.dtypes)

# Summary statistics
print("\nSummary Statistics:")
display(df.describe(include="all"))
```

```python
# Create a copy of the original dataset
df_clean = df.copy()

# Convert sale_date to datetime
df_clean["sale_date"] = pd.to_datetime(
    df_clean["sale_date"], errors="coerce"
)

# Standardize text columns
text_columns = ["product", "region"]

for col in text_columns:
    df_clean[col] = (
        df_clean[col]
        .str.strip()
        .str.title()
    )

# Check updated data types and values
print("Updated Data Types:")
print(df_clean.dtypes)

print("\nStandardized Text Values:")
display(df_clean[["product", "region"]])

print("\nMissing Values After Type Conversion:")
print(df_clean.isnull().sum())
```

```python
# Detect outliers using the IQR method
numeric_columns = ["amount", "quantity", "discount"]

outlier_summary = []

for col in numeric_columns:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[col] < lower_bound) |
        (df_clean[col] > upper_bound)
    ]

    outlier_summary.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers)
    })

print("IQR Outlier Detection:")
display(pd.DataFrame(outlier_summary))
```

```python
# Remove duplicate rows (none found in this dataset)
df_clean = df_clean.drop_duplicates()

# Save the cleaned dataset
output_path = Path("cleaned_dataset.csv")
df_clean.to_csv(output_path, index=False)

print("Cleaned dataset saved to:", output_path.resolve())
print("Final dataset shape:", df_clean.shape)
display(df_clean)
```
