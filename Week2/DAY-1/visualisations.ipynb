```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the sales dataset
df = pd.read_csv("north_sales.csv")

# Display the data
display(df)
```

```python
# Bar chart: Sales amount by product
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="product", y="amount")

plt.title("Sales Amount by Product")
plt.xlabel("Product")
plt.ylabel("Sales Amount (₹)")

plt.tight_layout()
plt.show()
```

```python
# Convert sale_date to datetime
df["sale_date"] = pd.to_datetime(df["sale_date"])

# Line chart: Sales amount over time
plt.figure(figsize=(8, 5))

sns.lineplot(
    data=df,
    x="sale_date",
    y="amount",
    marker="o"
)

plt.title("Sales Amount Over Time")
plt.xlabel("Sale Date")
plt.ylabel("Sales Amount (₹)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Scatter plot: Quantity vs Sales Amount
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="quantity",
    y="amount",
    hue="product",
    s=150
)

plt.title("Quantity vs Sales Amount")
plt.xlabel("Quantity Sold")
plt.ylabel("Sales Amount (₹)")

plt.tight_layout()
plt.show()
```

```python
# Histogram: Distribution of sales amounts
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="amount",
    bins=3,
    kde=False
)

plt.title("Distribution of Sales Amounts")
plt.xlabel("Sales Amount (₹)")
plt.ylabel("Number of Sales")

plt.tight_layout()
plt.show()
```

```python
# Box plot: Distribution of sales amounts
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="amount")

plt.title("Box Plot of Sales Amounts")
plt.xlabel("Sales Amount (₹)")

plt.tight_layout()
plt.show()
```

```python
# Bar chart with annotations
plt.figure(figsize=(8, 5))

ax = sns.barplot(data=df, x="product", y="amount")

plt.title("Sales Amount by Product")
plt.xlabel("Product")
plt.ylabel("Sales Amount (₹)")

# Add values above each bar
for container in ax.containers:
    ax.bar_label(container, fmt="₹%.0f", padding=3)

sns.set_theme(style="whitegrid")
plt.tight_layout()
plt.show()
```

```python
from pathlib import Path

# Create a folder to store the charts
output_dir = Path("charts")
output_dir.mkdir(exist_ok=True)

# Apply Seaborn styling
sns.set_theme(style="whitegrid")

# Define the five charts
chart_functions = [
    ("bar_chart.png", lambda: sns.barplot(data=df, x="product", y="amount")),
    ("line_chart.png", lambda: sns.lineplot(data=df, x="sale_date", y="amount", marker="o")),
    ("scatter_plot.png", lambda: sns.scatterplot(data=df, x="quantity", y="amount", hue="product")),
    ("histogram.png", lambda: sns.histplot(data=df, x="amount", bins=3)),
    ("box_plot.png", lambda: sns.boxplot(data=df, x="amount")),
]

# Save every chart as a high-quality PNG
for filename, plot_function in chart_functions:
    fig, ax = plt.subplots(figsize=(8, 5))
    plot_function()
    ax.set_title(filename.replace("_", " ").replace(".png", "").title())
    fig.tight_layout()
    fig.savefig(output_dir / filename, dpi=300, bbox_inches="tight")
    plt.close(fig)

print("All five charts saved successfully!")
```
