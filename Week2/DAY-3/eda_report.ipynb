```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("EDA Workflow Started!")
```

```python
df = pd.read_csv("../../north_sales.csv")

print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
display(df.head())
```

```python
print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nSummary Statistics:")
display(df.describe())
```

```python
df["sale_date"] = pd.to_datetime(df["sale_date"])

print(df.dtypes)
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="product", y="amount")
plt.title("Sales Amount by Product")
plt.xlabel("Product")
plt.ylabel("Sales Amount (₹)")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="region", y="amount")
plt.title("Sales Amount by Region")
plt.xlabel("Region")
plt.ylabel("Sales Amount (₹)")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.heatmap(
    df.select_dtypes(include="number").corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
