# Project 1: Indian E-Commerce Data Analysis

## Objective
To perform exploratory data analysis (EDA) on an Indian e-commerce dataset using Python, Pandas, and Matplotlib to identify five key business findings.

## Dataset
The dataset contains information about orders, customers, products, order items, and shipments. It is synthetically generated and may not represent actual real-world business performance.

## Tools Used
- Python
- Pandas
- Matplotlib
- Jupyter Notebook

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Dataset folder
DATA_DIR = Path("Data")

# Load the datasets
orders = pd.read_csv(DATA_DIR / "orders.csv")
customers = pd.read_csv(DATA_DIR / "customers.csv")
products = pd.read_csv(DATA_DIR / "products.csv")
order_items = pd.read_csv(DATA_DIR / "order_items.csv")
shipments = pd.read_csv(DATA_DIR / "shipments.csv")

print("All datasets loaded successfully!\n")

print("Orders:", orders.shape)
print("Customers:", customers.shape)
print("Products:", products.shape)
print("Order items:", order_items.shape)
print("Shipments:", shipments.shape)
```

```python
# Inspect the structure of each dataset

datasets = {
    "Orders": orders,
    "Customers": customers,
    "Products": products,
    "Order Items": order_items,
    "Shipments": shipments
}

for name, df in datasets.items():
    print(f"\n{'=' * 50}")
    print(name)
    print("Columns:", df.columns.tolist())
    print("\nFirst 3 rows:")
    display(df.head(3))
```

```python
# Check missing values and duplicate rows in each dataset

for name, df in datasets.items():
    print(f"\n{'=' * 45}")
    print(name)
    print("Total rows:", len(df))
    print("Duplicate rows:", df.duplicated().sum())
    print("\nMissing values:")
    print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# Missing-value analysis for all datasets

for name, df in datasets.items():
    missing = df.isnull().sum()
    missing = missing[missing > 0]

    print(f"\n{name}")
    if missing.empty:
        print("No missing values.")
    else:
        report = pd.DataFrame({
            "Missing Count": missing,
            "Missing Percentage": (missing / len(df) * 100).round(2)
        })
        display(report)
```

```python
# Create working copies so the original data stays unchanged

orders_clean = orders.copy()
customers_clean = customers.copy()
products_clean = products.copy()
order_items_clean = order_items.copy()
shipments_clean = shipments.copy()

# Convert date columns to datetime where available

date_columns = {
    "orders": ["order_date"],
    "customers": ["last_order_date"],
    "shipments": [
        "shipment_date",
        "expected_delivery_date",
        "actual_delivery_date"
    ]
}

cleaned_datasets = {
    "orders": orders_clean,
    "customers": customers_clean,
    "products": products_clean,
    "order_items": order_items_clean,
    "shipments": shipments_clean
}

for name, columns in date_columns.items():
    df = cleaned_datasets[name]

    for column in columns:
        if column in df.columns:
            df[column] = pd.to_datetime(df[column], errors="coerce")

print("Working copies created and available date columns converted.")
print("Original datasets remain unchanged.")
```

```python
# Summary statistics for numerical columns

print("ORDERS — Numerical Summary")
display(orders_clean.describe().round(2))

print("\nCUSTOMERS — Numerical Summary")
display(customers_clean.describe().round(2))

print("\nPRODUCTS — Numerical Summary")
display(products_clean.describe().round(2))

print("\nORDER ITEMS — Numerical Summary")
display(order_items_clean.describe().round(2))
```

```python
# Analyse order status distribution

print("Order status counts:")
display(orders_clean["order_status"].value_counts().to_frame("Number of Orders"))

# Visualise order status distribution
orders_clean["order_status"].value_counts().plot(
    kind="bar",
    figsize=(8, 5),
    title="Distribution of Order Status"
)

plt.xlabel("Order Status")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Analyse sales and profit by product category

print("Product columns:")
print(products_clean.columns.tolist())

print("\nOrder item columns:")
print(order_items_clean.columns.tolist())
```

```python
# Analyse revenue and profit by product category

category_analysis = order_items_clean.merge(
    products_clean[["product_id", "category"]],
    on="product_id",
    how="left"
)

category_summary = category_analysis.groupby("category").agg(
    Total_Revenue=("item_revenue", "sum"),
    Total_Profit=("profit", "sum"),
    Items_Sold=("quantity", "sum")
).sort_values("Total_Revenue", ascending=False)

display(category_summary.round(2))
```

```python
# Visualise revenue by product category

category_summary["Total_Revenue"].sort_values().plot(
    kind="barh",
    figsize=(9, 5),
    title="Total Revenue by Product Category"
)

plt.xlabel("Total Revenue")
plt.ylabel("Product Category")
plt.tight_layout()
plt.show()
```

```python
# Analyse customer segments

print("Customer columns:")
print(customers_clean.columns.tolist())
```

```python
# Analyse customer segments

segment_summary = customers_clean.groupby("customer_segment").agg(
    Customer_Count=("customer_id", "count"),
    Average_Spend=("total_spend", "mean"),
    Average_Order_Value=("average_order_value", "mean"),
    Average_Orders=("total_orders", "mean")
).sort_values("Average_Spend", ascending=False)

display(segment_summary.round(2))
```

```python
# Inspect shipment columns before analysing delivery performance

print(shipments_clean.columns.tolist())
```

```python
# Analyse delivery time by shipping method

delivery_summary = shipments_clean.groupby("shipping_method").agg(
    Average_Delivery_Days=("delivery_days", "mean"),
    Shipment_Count=("shipment_id", "count"),
    Delayed_Shipments=("delayed_flag", "sum")
)

delivery_summary["Delay_Rate_Percent"] = (
    delivery_summary["Delayed_Shipments"]
    / delivery_summary["Shipment_Count"] * 100
)

display(delivery_summary.round(2))
```

```python
# Visualise average delivery time by shipping method

delivery_summary["Average_Delivery_Days"].sort_values().plot(
    kind="bar",
    figsize=(9, 5),
    title="Average Delivery Time by Shipping Method"
)

plt.xlabel("Shipping Method")
plt.ylabel("Average Delivery Days")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Analyse monthly order value

orders_clean["order_date"] = pd.to_datetime(
    orders_clean["order_date"], errors="coerce"
)

monthly_sales = (
    orders_clean.dropna(subset=["order_date"])
    .set_index("order_date")
    .resample("MS")["final_amount"]
    .sum()
)

display(monthly_sales.to_frame("Total_Order_Value").head())

monthly_sales.plot(
    figsize=(10, 5),
    marker="o",
    title="Monthly Recorded Order Value"
)

plt.xlabel("Month")
plt.ylabel("Recorded Order Value")
plt.tight_layout()
plt.show()
```

```python
# Find the months with the highest recorded order value

monthly_summary = monthly_sales.to_frame(
    name="Total_Order_Value"
)

monthly_summary["Previous_Month"] = (
    monthly_summary["Total_Order_Value"].shift(1)
)

monthly_summary["Change_Percent"] = (
    (monthly_summary["Total_Order_Value"]
     - monthly_summary["Previous_Month"])
    / monthly_summary["Previous_Month"] * 100
)

display(
    monthly_summary.sort_values(
        "Total_Order_Value", ascending=False
    ).head(5).round(2)
)

print("\nLargest monthly percentage increase:")
display(
    monthly_summary.sort_values(
        "Change_Percent", ascending=False
    ).head(3).round(2)
)
```

## Business Insight Report — Five Key Findings

### 1. Order Status
80,679 out of 100,000 orders were marked as Delivered (80.68%). Cancelled, returned, and failed orders should be investigated to identify potential service and fulfilment improvements.

### 2. Product Category Performance
Electronics recorded the highest category revenue, followed by Appliances. The business could review inventory availability and profitability in these categories before making stocking decisions.

### 3. Customer Segmentation
Premium customers had the highest average recorded spend (₹286,622.67), while Regular customers formed the largest segment with 11,002 customers. Targeted loyalty initiatives and repeat-purchase campaigns could be tested.

### 4. Delivery Performance
Next-Day shipping averaged 1.81 delivery days and had a 10.60% delay rate. Standard shipping had 21,519 delayed shipments, while Same-Day shipping had a 56.40% delay rate. Delivery reliability should be investigated alongside shipping costs and service expectations.

### 5. Monthly Order Value
December 2025 had the highest recorded monthly order value, approximately ₹247.76 million. October 2025 recorded a 65.92% increase over September. These spikes should be validated before drawing conclusions about business growth.

## Recommendations

1. Investigate cancellations, returns, and failed orders to identify recurring causes.
2. Review category profitability and customer segments before changing inventory or marketing strategies.
3. Investigate delivery delays and validate unusual monthly order-value spikes.

## Limitations
The dataset is synthetically generated and may not reflect real-world business performance. Recorded order value is not necessarily realised revenue, particularly when cancelled or returned orders are included.
